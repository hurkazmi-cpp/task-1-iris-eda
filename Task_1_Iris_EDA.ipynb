# Task 1: Exploratory Data Analysis (EDA) on the Iris Dataset
## Problem Statement & Goal
The objective is to load, inspect, and visualize the Iris dataset to understand feature distributions and identify outliers before modeling.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Loading the data
data = sns.load_dataset("iris")
```

## Dataset Structure and Attributes
Before diving into statistical analysis, we first inspect the overall dimensions (shape) of the dataset and verify the names of the columns to understand our available features.

```python
# Inspection of data
print(f"Shape of data: {data.shape}")
print(f"Column names: {list(data.columns)}")
```

## Structural Integrity and Descriptive Statistics
Here, we preview the first 15 rows of the raw data to see the data layout. We then utilize the `.info()` and `.describe()` methods to check for missing values (if any), verify data types, and view fundamental summary statistics (such as mean, standard deviation, and quartiles).

```python
print("\n--- First 15 Rows --- ")
print(data.head(15))

print("\n--- Dataset Info --- ")
print(data.info())

print("\n--- Summary Statistics --- ")
print(data.describe())
```

### Visual Analysis 1: Scatter Plot
The purpose of this visualization is to analyze the relationship between `sepal_length` and `petal_length`. By applying a color hue based on the `species`, we can observe how the biological measurements correlate and whether the different species naturally separate into distinct clusters.

```python
# Data visualization
# Scatterplot
plt.figure(figsize=(8, 5))
sns.scatterplot(x="sepal_length", y="petal_length", data=data, hue="species")
plt.title("Sepal Length VS Petal Length across Iris Species")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.tight_layout()
plt.show()
```

### Visual Analysis 2: Histogram
This histogram maps the frequency and distribution of `sepal_length` measurements. Layering the data by species allows us to look at data density, determine if the features follow a normal (bell-curve) distribution, and see where each species' measurements typically center.

```python
# Histogram
plt.figure(figsize=(8, 5))
sns.histplot(x="sepal_length", data=data, hue="species", bins=10)
plt.title("Distribution of Sepal Length by Species")
plt.xlabel("Sepal Length (cm)")
plt.tight_layout()
plt.show()
```

### Visual Analysis 3: Box Plot
We utilize a box plot to examine the statistical spread of the data and isolate anomalies. By placing the categorical `species` on the X-axis and `sepal_length` on the Y-axis, we can easily identify the median, variance, and detect any data points sitting outside the whiskers as potential outliers.

```python
# Boxplot
plt.figure(figsize=(8, 5))
sns.boxplot(x="species", y="sepal_length", data=data, hue="species", palette="pastel")
plt.title("Sepal Length Distribution & Outliers by Species")
plt.xlabel("Species")
plt.ylabel("Sepal Length (cm)")
plt.tight_layout()
plt.show()
```
